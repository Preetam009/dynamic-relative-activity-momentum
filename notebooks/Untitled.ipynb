```python
import os
import numpy as np
import pandas as pd
from pathlib import Path

base_path = Path('../data')
symbols_path = Path('../watchlists/symbols.csv')
```

```python
symbols = pd.read_csv(symbols_path)
```

```python
def features(path:str, symbol:str) -> pd.DataFrame:
    df = pd.read_csv(path)
    df["TradedValue"] = df["Volume"] * df["Close"]
    df["return_3d"] = df["Close"].pct_change(3)
    df["return_6d"] = df["Close"].pct_change(6)
    df['Symbol'] = symbol

    return df
```

```python
data = pd.DataFrame()

for symbol in symbols['Symbol'].values:
    try:
        symbol = symbol.replace("&", "_AMP_").replace("-", "_DASH_")
        data = pd.concat([data, features(Path(os.path.join(base_path, symbol+'.csv')), symbol)])
        
    except: 
        print(f"Error occured at {symbol}")

data = data.sort_values(['Symbol', 'TimeStamp'])
```

```python
data.to_csv('Master_Data.csv')
```
